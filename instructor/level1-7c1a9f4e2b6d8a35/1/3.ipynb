## Pemeriksaan awal

Jalankan cell berikut untuk menyiapkan feedback otomatis.


```python
def check_equal(label, actual, expected):
    if actual == expected:
        print(f"OK: {label}")
    else:
        print(f"Belum tepat: {label}")
        print(f"  Diperoleh: {actual!r}")
        print(f"  Diharapkan: {expected!r}")


def check_call(label, function, args, expected):
    if not callable(function):
        print(f"Belum tepat: {label}. Function belum dibuat")
        return
    try:
        actual = function(*args)
    except Exception as error:
        print(f"Belum tepat: {label}. Pemanggilan menghasilkan {type(error).__name__}")
        return
    check_equal(label, actual, expected)


print("Pemeriksaan awal selesai. Python siap digunakan.")
```

### Latihan

Buat function `clean_names(raw_names)` yang membersihkan whitespace, memakai title case, dan melewati string kosong.


```python
def clean_names(raw_names):
    result = []
    for raw_name in raw_names:
        clean_name = raw_name.strip().title()
        if clean_name == "":
            continue
        result.append(clean_name)
    return result
```

```python
check_call("nama kosong", globals().get("clean_names"), ([],), [])
```

```python
check_call("satu nama", globals().get("clean_names"), (["  anisa "],), ["Anisa"])
```

```python
check_call("melewati teks kosong", globals().get("clean_names"), ([" ", "BUDI"],), ["Budi"])
```

### Latihan

Buat function `summarize_scores(scores)`. Untuk list kosong, kembalikan dictionary dengan count 0, total 0, lowest None, highest None. Untuk data lain, hitung dengan loop biasa.


```python
def summarize_scores(scores):
    if len(scores) == 0:
        return {"count": 0, "total": 0, "lowest": None, "highest": None}
    total = 0
    lowest = scores[0]
    highest = scores[0]
    for score in scores:
        total = total + score
        if score < lowest:
            lowest = score
        if score > highest:
            highest = score
    return {"count": len(scores), "total": total, "lowest": lowest, "highest": highest}
```

```python
check_call("ringkasan kosong", globals().get("summarize_scores"), ([],), {"count": 0, "total": 0, "lowest": None, "highest": None})
```

```python
check_call("ringkasan satu elemen", globals().get("summarize_scores"), ([75],), {"count": 1, "total": 75, "lowest": 75, "highest": 75})
```

```python
check_call("ringkasan banyak elemen", globals().get("summarize_scores"), ([78, 92, 60],), {"count": 3, "total": 230, "lowest": 60, "highest": 92})
```

```python
check_call("ringkasan duplikat", globals().get("summarize_scores"), ([75, 75, 75],), {"count": 3, "total": 225, "lowest": 75, "highest": 75})
```

### Latihan

Buat function `filter_passed(scores, minimum=75)` yang mengembalikan list nilai yang memenuhi batas. Gunakan loop biasa.


```python
def filter_passed(scores, minimum=75):
    result = []
    for score in scores:
        if score >= minimum:
            result.append(score)
    return result
```

```python
check_call("filter kosong", globals().get("filter_passed"), ([],), [])
```

```python
check_call("filter batas", globals().get("filter_passed"), ([74, 75, 76],), [75, 76])
```

```python
check_call("filter duplikat", globals().get("filter_passed"), ([75, 60, 75],), [75, 75])
```

### Latihan

Buat function `find_target(items, target)` yang mengembalikan index pertama target atau `-1` jika tidak ditemukan.


```python
def find_target(items, target):
    for index in range(len(items)):
        if items[index] == target:
            return index
    return -1
```

```python
check_call("target tidak ada", globals().get("find_target"), (["A", "B"], "C"), -1)
```

```python
check_call("target pertama", globals().get("find_target"), (["A", "B", "A"], "A"), 0)
```

```python
check_call("target terakhir", globals().get("find_target"), (["A", "B"], "B"), 1)
```

### Latihan

Buat function `make_pairs(left_items, right_items)` yang mengembalikan list string `left-right` untuk setiap pasangan menggunakan nested loop.


```python
def make_pairs(left_items, right_items):
    pairs = []
    for left in left_items:
        for right in right_items:
            pairs.append(f"{left}-{right}")
    return pairs
```

```python
check_call("nested loop kosong", globals().get("make_pairs"), (["A"], []), [])
```

```python
check_call("nested loop pasangan", globals().get("make_pairs"), (["A", "B"], [1, 2]), ["A-1", "A-2", "B-1", "B-2"])
```
