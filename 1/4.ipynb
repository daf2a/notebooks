## Contoh terbimbing / Guided example

```python
def calculate_score(correct, wrong):
    return 4 * correct - wrong

participants = [{'name': 'Alya', 'correct': 17, 'wrong': 3}, {'name': 'Bima', 'correct': 20, 'wrong': 4}]
for person in participants:
    person['score'] = calculate_score(person['correct'], person['wrong'])
print(participants)
```

## Challenge
Cari peserta terbaik dan pastikan input kosong ditangani. / Find the best participant and safely handle an empty input.

```python
def best_participant(records):
    if not records:
        return None
    best = records[0]
    for record in records[1:]:
        if record['score'] > best['score']:
            best = record
    return best

assert best_participant([]) is None
assert best_participant(participants)['name'] == 'Bima'
print(best_participant(participants))
```

## Pembahasan dan solusi / Discussion and solutions
Fungsi perhitungan mengembalikan nilai agar dapat diuji. Guard clause menangani list kosong sebelum indeks 0 diakses. / Calculation functions return values for testing. A guard clause handles an empty list before index 0 is accessed.
