## Pemeriksaan awal

Jalankan cell berikut untuk menyiapkan feedback otomatis.


````python
def check_equal(label, actual, expected):
    if actual == expected:
        print(f"OK: {label}")
    else:
        print(f"Belum tepat: {label}")
        print(f"  Diperoleh: {actual!r}")
        print(f"  Diharapkan: {expected!r}")


def check_call(label, function, args, expected):
    if not callable(function):
        print(f"Belum tepat: {label}. Function belum dibuat")
        return
    try:
        actual = function(*args)
    except Exception as error:
        print(f"Belum tepat: {label}. Pemanggilan menghasilkan {type(error).__name__}")
        return
    check_equal(label, actual, expected)


print("Pemeriksaan awal selesai. Python siap digunakan.")
````

## Operasi aritmetika

Setelah data type, bandingkan pembagian biasa, pembagian bulat, sisa bagi, pangkat, dan prioritas operator.


````python
left = 17
right = 5
print(left + right)
print(left - right)
print(left * right)
print(left / right)
print(left // right)
print(left % right)
print(2 ** 3)
print(4 + 3 * 2)
print((4 + 3) * 2)
````

## Type conversion


````python
quantity_text = "6"
quantity = int(quantity_text)
price = float("7500.50")
print(quantity_text * 2)
print(quantity * 2)
print(f"Harga: Rp{price:.2f}")
````
