## Challenge
Buat kategori A/B/C/D dari batas 90/80/70. / Build A/B/C/D categories using 90/80/70 boundaries.

```python
def grade(value):
    if value >= 90: return 'A'
    if value >= 80: return 'B'
    if value >= 70: return 'C'
    return 'D'
assert [grade(x) for x in [95, 85, 75, 65]] == ['A', 'B', 'C', 'D']
```
