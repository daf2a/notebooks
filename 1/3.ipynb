## Contoh terbimbing / Guided example

```python
scores = [72, 88, 91, 65, 84]
running = 0
for score in scores:
    running += score
    print('accumulator:', running)
print('mean:', running / len(scores))
```

## Challenge
Bersihkan nama, lalu hitung jumlah pasangan pada loop bersarang. / Clean names, then count nested-loop pairs.

```python
raw = ['  alya', 'BIMA  ', ' ciTra ']
clean = [name.strip().title() for name in raw]
pairs = 0
for left in clean:
    for right in clean:
        pairs += 1
print(clean, pairs)
assert pairs == len(clean) ** 2
```

## Pembahasan dan solusi / Discussion and solutions
Accumulator bertambah hanya saat kondisi benar. Dua loop atas tiga nama memeriksa $3^2=9$ pasangan. / The accumulator changes only when the condition is true. Two loops over three names inspect $3^2=9$ pairs.
